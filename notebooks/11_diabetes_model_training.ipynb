```python
import pandas as pd
import numpy as np
import os
import joblib

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier
```

```python
X_train = pd.read_csv("../datasets/processed/diabetes/x_train.csv")
X_test = pd.read_csv("../datasets/processed/diabetes/x_test.csv")

y_train = pd.read_csv("../datasets/processed/diabetes/y_train.csv").squeeze()
y_test = pd.read_csv("../datasets/processed/diabetes/y_test.csv").squeeze()

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    
    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),
    
    "Random Forest": RandomForestClassifier(
        random_state=42
    ),
    
    "SVM": SVC(
        probability=True,
        random_state=42
    ),
    
    "XGBoost": XGBClassifier(
        random_state=42
    )
}

print("Models created:")
for name in models:
    print("-", name)
```

```python
os.makedirs("../models/diabetes", exist_ok=True)

print("Diabetes model folder ready.")
```

```python
trained_models = {}

for name, model in models.items():
    print(f"Training {name}...")
    
    model.fit(X_train, y_train)
    trained_models[name] = model
    
    print(f"{name} training completed.")
```

```python
for name, model in trained_models.items():
    
    file_name = name.lower().replace(" ", "_") + ".pkl"
    file_path = "../models/diabetes/" + file_name
    
    joblib.dump(model, file_path)
    
    print("Saved:", file_path)
```

```python
best_diabetes_model = trained_models["Decision Tree"]

joblib.dump(
    best_diabetes_model,
    "../models/diabetes/best_model.pkl"
)

print("Best Diabetes model saved successfully!")
print("Model: Decision Tree")
```
