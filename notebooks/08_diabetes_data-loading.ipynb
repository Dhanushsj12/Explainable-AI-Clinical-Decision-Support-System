```python
import pandas as pd
import numpy as np

# Path to the raw diabetes dataset
DATA_PATH = "../datasets/raw/diabetes_data_upload.csv"

# Load dataset
df = pd.read_csv(DATA_PATH)

# Dataset shape
print("Dataset Shape:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Data types
print("\nData Types:")
print(df.dtypes)

# First 5 rows
print("\nFirst 5 Rows:")
display(df.head())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Target distribution
print("\nTarget Distribution:")
print(df["class"].value_counts())
```

```python
for column in df.columns:
    print(f"\n{column}")
    print(df[column].unique())
```

```python
# Identify categorical columns

categorical_columns = df.select_dtypes(include=["str", "object"]).columns

print("Categorical Columns:")
print(categorical_columns.tolist())

print("\nNumber of categorical columns:", len(categorical_columns))
```

```python
print("Visual blurring values:")
print(df["visual blurring"].value_counts())

print("\nMuscle stiffness values:")
print(df["muscle stiffness"].value_counts())

print("\nAlopecia values:")
print(df["Alopecia"].value_counts())

print("\nObesity values:")
print(df["Obesity"].value_counts())
```

```python
# Create a copy so the original raw dataset is not modified
diabetes_cleaned = df.copy()

# Convert Gender
diabetes_cleaned["Gender"] = diabetes_cleaned["Gender"].map({
    "Female": 0,
    "Male": 1
})

# Convert Yes/No clinical features
yes_no_columns = [
    "Polyuria",
    "Polydipsia",
    "sudden weight loss",
    "weakness",
    "Polyphagia",
    "Genital thrush",
    "visual blurring",
    "Itching",
    "Irritability",
    "delayed healing",
    "partial paresis",
    "muscle stiffness",
    "Alopecia",
    "Obesity"
]

for column in yes_no_columns:
    diabetes_cleaned[column] = diabetes_cleaned[column].map({
        "No": 0,
        "Yes": 1
    })

# Convert target
diabetes_cleaned["class"] = diabetes_cleaned["class"].map({
    "Negative": 0,
    "Positive": 1
})

print("Encoded dataset:")
display(diabetes_cleaned.head())

print("\nData Types:")
print(diabetes_cleaned.dtypes)
```

```python
# Verify that all values are numeric

print("Missing values after encoding:")
print(diabetes_cleaned.isnull().sum())

print("\nData types after encoding:")
print(diabetes_cleaned.dtypes)

print("\nTarget distribution:")
print(diabetes_cleaned["class"].value_counts())
```

```python
# Save the cleaned diabetes dataset

CLEANED_PATH = "../datasets/cleaned/diabetes_cleaned.csv"

diabetes_cleaned.to_csv(CLEANED_PATH, index=False)

print("Cleaned diabetes dataset saved successfully!")
print("Location:", CLEANED_PATH)
print("Shape:", diabetes_cleaned.shape)
```
