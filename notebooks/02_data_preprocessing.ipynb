```python
# ==========================================
# HEART DISEASE DATA PREPROCESSING
# ==========================================

# ==========================================
# Step 1: Import Libraries
# ==========================================

import pandas as pd
import numpy as np

# ==========================================
# Step 2: Load Cleaned Dataset
# ==========================================

df = pd.read_csv("../datasets/cleaned/heart_cleaned.csv")

print("Dataset Loaded Successfully!")

# ==========================================
# Step 3: Display Dataset
# ==========================================

print("\nFirst 5 Rows:")
print(df.head())

# ==========================================
# Step 4: Dataset Shape
# ==========================================

print("\nDataset Shape:")
print(df.shape)

# ==========================================
# Step 5: Dataset Information
# ==========================================

print("\nDataset Information:")
df.info()

# ==========================================
# Step 6: Check Missing Values
# ==========================================

print("\nMissing Values:")
print(df.isnull().sum())

# ==========================================
# Step 7: Check Duplicate Rows
# ==========================================

duplicates = df.duplicated().sum()

print("\nDuplicate Rows:", duplicates)

# Remove duplicates if any
df.drop_duplicates(inplace=True)

# ==========================================
# Step 8: Convert Target into Binary
# ==========================================

# 0 = No Heart Disease
# 1,2,3,4 = Heart Disease

df["target"] = df["target"].apply(lambda x: 0 if x == 0 else 1)

print("\nTarget Distribution:")
print(df["target"].value_counts())

# ==========================================
# Step 9: Verify Data Types
# ==========================================

print("\nData Types:")
print(df.dtypes)

# ==========================================
# Step 10: Save Preprocessed Dataset
# ==========================================

output_path = "../datasets/processed/heart_preprocessed.csv"

df.to_csv(output_path, index=False)

print("\nPreprocessed Dataset Saved Successfully!")

print(output_path)

# ==========================================
# Step 11: Preview Dataset
# ==========================================

print("\nPreview:")
print(df.head())
```
