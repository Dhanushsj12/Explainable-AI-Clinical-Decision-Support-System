```python
# ==========================================
# Step 1: Import Libraries
# ==========================================

import pandas as pd
import numpy as np
import os

# ==========================================
# Step 2: Check Current Working Directory
# ==========================================

print("Current Working Directory:")
print(os.getcwd())

# ==========================================
# Step 3: Define Column Names
# ==========================================

column_names = [
    'age',
    'sex',
    'cp',
    'trestbps',
    'chol',
    'fbs',
    'restecg',
    'thalach',
    'exang',
    'oldpeak',
    'slope',
    'ca',
    'thal',
    'target'
]

# ==========================================
# Step 4: Load Dataset
# ==========================================

df = pd.read_csv(
    "../datasets/raw/processed.cleveland.data",
    header=None,
    names=column_names
)

print("\nDataset Loaded Successfully!")

# ==========================================
# Step 5: Display First 5 Rows
# ==========================================

print("\nFirst 5 Rows:")
print(df.head())

# ==========================================
# Step 6: Dataset Information
# ==========================================

print("\nDataset Information:")
df.info()

# ==========================================
# Step 7: Dataset Shape
# ==========================================

print("\nDataset Shape:")
print(df.shape)

# ==========================================
# Step 8: Statistical Summary
# ==========================================

print("\nStatistical Summary:")
print(df.describe())

# ==========================================
# Step 9: Replace '?' with NaN
# ==========================================

df.replace('?', np.nan, inplace=True)

# ==========================================
# Step 10: Convert Columns to Numeric
# ==========================================

df['ca'] = pd.to_numeric(df['ca'])
df['thal'] = pd.to_numeric(df['thal'])

# ==========================================
# Step 11: Check Missing Values
# ==========================================

print("\nMissing Values:")
print(df.isnull().sum())

# ==========================================
# Step 12: Fill Missing Values
# ==========================================

df['ca'] = df['ca'].fillna(df['ca'].median())
df['thal'] = df['thal'].fillna(df['thal'].median())

# ==========================================
# Step 13: Check Duplicate Rows
# ==========================================

duplicates = df.duplicated().sum()

print("\nDuplicate Rows:", duplicates)

df.drop_duplicates(inplace=True)

# ==========================================
# Step 14: Final Dataset Information
# ==========================================

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# ==========================================
# Step 15: Save Cleaned Dataset
# ==========================================

output_path = "../datasets/cleaned/heart_cleaned.csv"

df.to_csv(
    output_path,
    index=False
)

print("\nCleaned Dataset Saved Successfully!")

print("\nSaved Location:")
print(output_path)

# ==========================================
# Step 16: Display Cleaned Dataset
# ==========================================

print("\nCleaned Dataset Preview:")
print(df.head())
```
