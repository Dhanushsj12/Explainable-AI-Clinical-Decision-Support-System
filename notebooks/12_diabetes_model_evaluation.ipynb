```python
import pandas as pd
import numpy as np
import joblib

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
X_test = pd.read_csv("../datasets/processed/diabetes/x_test.csv")
y_test = pd.read_csv("../datasets/processed/diabetes/y_test.csv").squeeze()

print("Test data:", X_test.shape)
print("Test target:", y_test.shape)
```

```python
models = {
    "Logistic Regression": joblib.load(
        "../models/diabetes/logistic_regression.pkl"
    ),
    
    "Decision Tree": joblib.load(
        "../models/diabetes/decision_tree.pkl"
    ),
    
    "Random Forest": joblib.load(
        "../models/diabetes/random_forest.pkl"
    ),
    
    "SVM": joblib.load(
        "../models/diabetes/svm.pkl"
    ),
    
    "XGBoost": joblib.load(
        "../models/diabetes/xgboost.pkl"
    )
}

print("All Diabetes models loaded successfully.")
```

```python
results = []

for name, model in models.items():
    
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]
    
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_prob)
    
    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": roc_auc
    })
```

```python
results_df = pd.DataFrame(results)

display(
    results_df.round(4)
)
```
