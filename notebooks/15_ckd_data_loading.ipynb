```python
import os
import re
import pandas as pd
import numpy as np
```

```python
BASE_DIR = os.path.abspath(
    os.path.join(os.getcwd(), "..")
)

CKD_RAW_PATH = os.path.join(
    BASE_DIR,
    "datasets",
    "raw",
    "ckd",
    "chronic_kidney_disease.arff"
)

print("CKD file:")
print(CKD_RAW_PATH)

print("\nFile exists:", os.path.exists(CKD_RAW_PATH))
```

```python
with open(CKD_RAW_PATH, "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Total lines:", len(lines))
```

```python
attributes = []

for line in lines:
    line = line.strip()

    if line.lower().startswith("@attribute"):
        parts = line.split(None, 2)

        if len(parts) >= 3:
            column_name = parts[1].strip("'\"")
            attributes.append(column_name)

print("Number of attributes:", len(attributes))
print("\nColumns:")
print(attributes)
```

```python
data_start = None

for i, line in enumerate(lines):
    if line.strip().lower() == "@data":
        data_start = i + 1
        break

print("Data starts at line:", data_start)
```

```python
data_rows = []

for line in lines[data_start:]:
    line = line.strip()

    if not line:
        continue

    values = [value.strip() for value in line.split(",")]

    if len(values) == len(attributes):
        data_rows.append(values)

df = pd.DataFrame(data_rows, columns=attributes)

print("Dataset shape:", df.shape)
```
