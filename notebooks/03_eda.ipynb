```python
# ==========================================
# EXPLORATORY DATA ANALYSIS (EDA)
# ==========================================

# ==========================================
# Step 1: Import Libraries
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Style
sns.set_style("whitegrid")

# ==========================================
# Step 2: Load Dataset
# ==========================================

df = pd.read_csv("../datasets/processed/heart_preprocessed.csv")

print("Dataset Loaded Successfully!\n")

# ==========================================
# Step 3: Dataset Overview
# ==========================================

print("Dataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nSummary Statistics:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

# ==========================================
# Step 4: Class Distribution
# ==========================================

plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="target",
    hue="target",
    palette="Set2",
    legend=False
)

plt.title("Heart Disease Class Distribution")
plt.xlabel("Target")
plt.ylabel("Count")

plt.savefig("../reports/figures/class_distribution.png")

plt.show()

# ==========================================
# Step 5: Histograms
# ==========================================

df.hist(
    figsize=(15,12),
    bins=20
)

plt.tight_layout()

plt.savefig("../reports/figures/histograms.png")

plt.show()

# ==========================================
# Step 6: Correlation Heatmap
# ==========================================

plt.figure(figsize=(12,10))

sns.heatmap(
    df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.savefig("../reports/figures/correlation_heatmap.png")

plt.show()

# ==========================================
# Step 7: Boxplots
# ==========================================

plt.figure(figsize=(15,8))

df.boxplot()

plt.xticks(rotation=45)

plt.title("Boxplots of Features")

plt.savefig("../reports/figures/boxplots.png")

plt.show()

# ==========================================
# Step 8: Pairplot (Optional)
# ==========================================

sns.pairplot(
    df,
    hue="target",
    vars=["age","chol","thalach","oldpeak"]
)

plt.savefig("../reports/figures/pairplot.png")

plt.show()

print("\nEDA Completed Successfully!")
```
